```python
import json
from typing import Any

class Agent:    
    def __init__(self, client: InferenceClient, system: str ="", tools: list = None) -> None:
        '''Why we initialize the default value of tpols param is 'None'?
            Bcs list obj is muttabele, This the explanation : 
            if we assign [] which is empty list on tools param, when we initialize 2 object it goes to the same tools
            param object, which is the same memory. And if we're going to .append either, the another one will also have
            the same value, bcs both are refers to the same memory.  
        '''
        self.client = client
        self.system = system    # tidak mengisi nones seperti tools karena string immutable
        self.messages = []
        self.tools = tools if tools is not None else[]
        if self.system:
            self.messages.append({"role":"system", "content":system})


    def __call__(self, message=""):
        if message: #for check message variable has a value, which is string
            self.messages.append({"role" : "user", "content" : message})

        final_assistant_content = self.execute()    #for access class method, in this case execute method/function
        if final_assistant_content:
            self.messages.append({"role" : "assistant", "content" : final_assistant_content})

        return final_assistant_content
    

    def execute(self):
        while True: #  Agentic reasoning loop - keep calling agent until it returns a response (not a tool call)
            completion = self.client.chat.completions.create(
                messages=self.messages,
                tools=self.tools,
                tool_choice="auto"
            )

            response_message = completion.choices[0].message
            if response_message.tool_calls:
                self.messages.append(response_message)

                tool_outputs = []
                for tool_call in response_message.tool_calls:   # process multiple tools
                    function_name = tool_call.function.name
                    function_args = json.loads(tool_call.function.arguments)

                    # Execute the Tools/Functions
                    if function_name in globals() and callable(globals()[function_name]):
                        function_to_call = globals()[function_name]
                        executed_output = function_to_call(**function_args)
                        tool_output_content = str(executed_output)  #ensure output string
                        print(f"Executing tool: {function_name} with args {function_args}, Output: {tool_output_content[:500]}...") #Debug print
                              
                    tool_outputs.append(
                        {
                            "tool_call_id": tool_call.id,
                            "role": "tool",
                            "name": function_name,
                            "content": tool_output_content,
                        }
                    )
                
                self.messages.extend(tool_outputs)  #use extend because tool_outputs is a list object, if we use append its going to nested list in a messages list
            else:
                print(response_message.content)
                return response_message.content

    # def execute2 (self):
    #     completion = self.client.chat.completions.create(
    #         message = self.message,
    #         tools=self.tools,
    #         tool_choice="auto",
    #     )
    #     response_message = completion.choices[0].message
    #     while response_message.tool_calls:
    #         self.messages.append(completion)
            
    #         tools_output = []
    #         for tool in response_message.tool_calls:
    #             function_neme = tool.function.name
    #             function_argument = tool.function.arguments
```

```python
from typing import Any
class C:
    def __init__(self, name, data =[]):
        self.name = name
        self.data = data if data is not None else []


    def __call__(self, x):
        return f"{self.name} called with {x}"

    def get_blabala(self):
        test = self.get_data()
        return test
    
    def get_data(self):
        return f"Data instance: {self.data}"

c = C("obj")
print(c("hello"))        # uses __call__ implicitly
print(c.get_blabala())   # explicit method call
# print(c.__str__)          

a = C("objeka")
print(id(a.data))       # uses __repr__/__str__ if defined
print(id(c.data))       # uses __repr__/__str__ if defined
```
